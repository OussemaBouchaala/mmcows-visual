```python
# Cell 9: Project configuration
from pathlib import Path

# === PATHS ===
DATA_ROOT = Path("/home/oussema/pfa-mmcows/store/data/raw/visual_data")
WORK_DIR  = Path("/home/oussema/mmcows-visual")  # your test/notebook repo

IMAGES    = DATA_ROOT / "images" / "0725"         # cam_1..cam_4
LABELS    = DATA_ROOT / "labels" / "combined" / "0725"  # YOLO txt files
PROJ_MAT  = DATA_ROOT / "proj_mat" / "0725"       # projection matrices
VIS_LOC   = DATA_ROOT / "visual_location"         # 3D coords per cow
BEH_LABELS= DATA_ROOT / "behavior_labels" / "individual"  # behavior CSVs

CAMERAS   = ["cam_1", "cam_2", "cam_3", "cam_4"]

# === VERIFY ===
for name, p in [("Images", IMAGES), ("Labels", LABELS), ("ProjMat", PROJ_MAT), ("VisLoc", VIS_LOC), ("Behavior", BEH_LABELS)]:
    exists = "✅" if p.exists() else "❌"
    print(f"{exists} {name}: {p}")

# Count matches
for cam in CAMERAS:
    n_img = len(list((IMAGES / cam).glob("*.jpg")))
    n_lbl = len(list((LABELS / cam).glob("*.txt")))
    print(f"  {cam}: {n_img} images, {n_lbl} labels")
```

```python
# Cell 10: Build master index + split
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split

IMAGES = Path("/home/oussema/pfa-mmcows/store/data/raw/visual_data/images/0725")
LABELS = Path("/home/oussema/pfa-mmcows/store/data/raw/visual_data/labels/combined/0725")
CAMERAS = ["cam_1", "cam_2", "cam_3", "cam_4"]

# Build index: one row per image
rows = []
for cam in CAMERAS:
    for img_path in sorted((IMAGES / cam).glob("*.jpg")):
        lbl_path = LABELS / cam / (img_path.stem + ".txt")
        rows.append({
            "camera": cam,
            "timestamp": img_path.stem.split("_")[0],  # e.g. 1690271846
            "image_path": str(img_path),
            "label_path": str(lbl_path),
            "has_label": lbl_path.exists()
        })

df = pd.DataFrame(rows)

# Split by TIMESTAMP (not random) so same moment doesn't leak across splits
# All 4 cameras at timestamp T go to the SAME split
unique_ts = sorted(df["timestamp"].unique())
ts_train, ts_temp = train_test_split(unique_ts, test_size=0.30, random_state=42)
ts_val, ts_test   = train_test_split(ts_temp,   test_size=0.50, random_state=42)

df["split"] = df["timestamp"].apply(
    lambda t: "train" if t in set(ts_train) else ("val" if t in set(ts_val) else "test")
)

print(df["split"].value_counts())
print(f"\nTotal: {len(df)} samples")
print(f"Unique timestamps: {len(unique_ts)}")
print(df.head())
```

```python
# Cell 12: Remap labels to single class + build YOLO dataset structure
import shutil
from pathlib import Path

LABELS_SRC = Path("/home/oussema/pfa-mmcows/store/data/raw/visual_data/labels/combined/0725")
IMAGES_SRC = Path("/home/oussema/pfa-mmcows/store/data/raw/visual_data/images/0725")
YOLO_ROOT  = Path("/home/oussema/pfa-mmcows/store/data/clean/yolo_dataset")

# Create YOLO folder structure
for split in ["train", "val", "test"]:
    (YOLO_ROOT / "images" / split).mkdir(parents=True, exist_ok=True)
    (YOLO_ROOT / "labels" / split).mkdir(parents=True, exist_ok=True)

count = {"train": 0, "val": 0, "test": 0}

for _, row in df.iterrows():
    split = row["split"]
    cam = row["camera"]
    fname = Path(row["image_path"]).stem
    
    # Symlink image (saves disk space)
    img_src = Path(row["image_path"])
    img_dst = YOLO_ROOT / "images" / split / f"{cam}_{fname}.jpg"
    if not img_dst.exists():
        img_dst.symlink_to(img_src)
    
    # Remap labels: change class ID → 0, keep bbox
    lbl_src = Path(row["label_path"])
    lbl_dst = YOLO_ROOT / "labels" / split / f"{cam}_{fname}.txt"
    if not lbl_dst.exists():
        with open(lbl_src) as f:
            lines = f.readlines()
        with open(lbl_dst, "w") as f:
            for line in lines:
                parts = line.strip().split()
                parts[0] = "0"  # all cows → class 0
                f.write(" ".join(parts) + "\n")
    
    count[split] += 1

print("✅ YOLO dataset built:")
for s, n in count.items():
    print(f"  {s}: {n} images")
```
